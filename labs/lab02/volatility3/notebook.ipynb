Usa este notebook para analizar la imagen de memoria del incidente.

```python
import subprocess, os
MEMORY = '/memory/incident.raw'
if os.path.exists(MEMORY):
    print('Imagen de memoria encontrada:', MEMORY)
else:
    print('⚠️ Coloca el archivo incident.raw en el directorio forensics/')
```

```python
# Listar procesos
result = subprocess.run(['vol', '-f', MEMORY, 'windows.pslist'], capture_output=True, text=True)
print(result.stdout[:3000])
```

```python
# Buscar strings sospechosos
result = subprocess.run(['vol', '-f', MEMORY, 'windows.strings', '--pid', '0'], capture_output=True, text=True)
for line in result.stdout.splitlines():
    if 'FLAG{' in line or 'password' in line.lower():
        print(line)
```
